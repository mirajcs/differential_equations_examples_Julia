# Second-Order Linear Differential Equations

This chapter covers equations of the form

$$y'' + p(t)\,y' + q(t)\,y = g(t),$$

which describe mechanical vibrations, electric circuits and many other systems where an acceleration depends on position and velocity. Following Boyce & DiPrima, *Elementary Differential Equations and Boundary Value Problems* (11th ed., Wiley), Chapter 3, the independent variable is now called $t$.

Prerequisite: [First-Order Differential Equations](../first-order-differential-equations.ipynb).
